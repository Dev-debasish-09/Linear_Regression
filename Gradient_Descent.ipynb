```python
import numpy as np
```

```python
# formation of gradient descent class
class GDregress:
  def __init__(self,learning_rate,iterations):
    self.learning_rate=learning_rate
    self.iterations=iterations
    self.m=27
    self.b=5
  def fit(self,x,y):
    for i in range(self.iterations):
      loss_slope_b =-2*np.sum( y - (self.m * x.ravel()) - self.b )
      loss_slope_m = -2*np.sum((y - self.m*x.ravel() - self.b)*x.ravel())
      self.b = self.b - (self.learning_rate * loss_slope_b)
      self.m = self.m - (self.learning_rate * loss_slope_m)
  def predict(self,x):
    return self.m*x+self.b
```

```python
# formation of batch Gradient Descent class
class batchGDregress:
  def __init__(self,learning_rate = 0.01,epochs = 10):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef = None
    self.intercept = None
  def fit(self,x,y):
    self.intercept = 0
    self.coef = np.ones(x.shape[0])
    for i in self.epochs :
      y_hat = np.dot(x,self.coef) + self.intercept
      intercept_derivative = -2*np.mean(y-y_hat)
      coef_derivative = -2*np.dot((y-y_hat),x)/x.shape[0]
      self.coef = self.coef - (self.learning_rate * coef_derivative)
      self.intercept = self.intercept - (self.learning_rate*intercept_derivative)
  def predict(self,x):
    return np.dot(x,self.coef) + self.intercept
```

```python
# formation of stocastic Gradient Descent class
class SGDregress:
  def __init__(self,learning_rate = 0.01,epochs = 10):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef = None
    self.intercept = None
  def fit(self,x,y):
    self.intercept = 0
    self.coef = np.ones(x.shape[0])
    for i in self.epochs :
      for j in range (x.shape[0]):
        idx = np.random.randint(0,x.shape[0])
        y_hat = np.dot(x[idx],self.coef) + self.intercept
        intercept_derivative = -2*(y[idx] - y_hat)
        coef_derivative = -2*np.dot((y[idx]-y_hat),x[idx])
        self.coef = self.coef - (self.learning_rate * coef_derivative)
        self.intercept = self.intercept - (self.learning_rate*intercept_derivative)
  def predict(self,x):
    return np.dot(x,self.coef) + self.intercept
```

```python
# formation of mini batch Gradient Descent class
class SGDregress:
  def __init__(self,batch_size , learning_rate = 0.01,epochs = 10):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coef = None
    self.intercept = None
    self.batch_size = batch_size
  def fit(self,x,y):
    self.intercept = 0
    self.coef = np.ones(x.shape[0])
    for i in self.epochs :
      for j in range (x.shape[0]/self.batch_size):
        idx = np.random.randint(0,x.shape[0])
        y_hat = np.dot(x[idx],self.coef) + self.intercept
        intercept_derivative = -2*(y[idx] - y_hat)
        coef_derivative = -2*np.dot((y[idx]-y_hat),x[idx])
        self.coef = self.coef - (self.learning_rate * coef_derivative)
        self.intercept = self.intercept - (self.learning_rate*intercept_derivative)
  def predict(self,x):
    return np.dot(x,self.coef) + self.intercept
```
